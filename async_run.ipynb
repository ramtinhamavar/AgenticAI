```python
import time
```

```python
def is_prime(n): # a bad way to determine if a number is prime (on purpose)
    prime = True
    for index in range(2, n):
        if n%index == 0:
            prime = False

    return prime

def is_perfect_square(n):
    perfect_square = False
    for i in range(1, n + 1):
        if i * i == n:
            perfect_square = True

    return perfect_square
```

```python
n = 10000000
tic = time.time()
prime_n = is_prime(n)
perfect_square_n = is_perfect_square(n)
toc = time.time()


print(f"{n} is prime?: {prime_n}")
print(f"{n} is perfect square?: {perfect_square_n}")

print(f"Time to calculate: {toc - tic}")
```

## Add Async

```python
import asyncio
```

```python
async def is_prime(n): # add async in front of functions
    prime = True
    for index in range(2, n):
        if n%index == 0:
            prime = False

    return prime

async def is_perfect_square(n):
    perfect_square = False
    for i in range(1, n + 1):
        if i * i == n:
            perfect_square = True

    return perfect_square
```

```python
# call them by await

n = 10000000
tic = time.time()
await asyncio.gather(is_prime(n), is_perfect_square(n))
toc = time.time()


print(f"{n} is prime?: {prime_n}")
print(f"{n} is perfect square?: {perfect_square_n}")

print(f"Time to calculate: {toc - tic}")
```
