```python
from pydantic import BaseModel
from typing import List, Dict, Optional
from pydantic import Field
from typing import Annotated # to add meta-data
```

```python
# define the schema!
class SubjectData(BaseModel):
    name: str
    folder_addr: str
    number_of_lost_teeth: int

    annotators: List[str]
    opg_center: Dict[str, str]

    anchor_addr: Optional[List[str]] = None

    family: str = Field(max_length=50)
```

```python
def add_subject_data(subject: SubjectData):
    print(f"Subject name: {subject.name}")
    print(f"Subject folder addr: {subject.folder_addr}")
    print(f"Subject number of lost teeth: {subject.number_of_lost_teeth}")
    print(f"List of annotators: {subject.annotators}")
    print(f"opg center name: {subject.opg_center.keys()}, opg center code: {subject.opg_center.values()}")
```

```python
subject_data = {"name":"Ramtin", "folder_addr":"/mnt/dev0/ramtinhamavar", "number_of_lost_teeth":2,
               "annotators":["annot_37", "annot_21"], "opg_center":{"Dr. Zhivago":"021"}, "family":"Hamavar"}
subject = SubjectData(**subject_data)

add_subject_data(subject)
```

## Special Validation

```python
from pydantic import field_validator
from pydantic import EmailStr
```

```python
class Subject(BaseModel):
    name: str
    email: EmailStr

    @field_validator("email")
    @classmethod
    def email_validator(cls, value):
        valid_domains = ["modares.ac.ir", "conc.can"]
        
        domain_name = value.split("@")[-1]
        if domain_name not in valid_domains:
            raise ValueError("not a valid domain")

        return value

    @field_validator("name")
    @classmethod
    def transform_name(cls, value):
        return value.upper()
```

```python
subject_dict = {"email":"r.hamavar@modares.ac.ir", "name":"Ramtin"}
subject = Subject(**subject_dict)
```

```python
print(f"Subject name: {subject.name}")
print(f"Subject email: {subject.email}")
```

## Model Validator - allows us to handle multiple fields

```python
from pydantic import model_validator

@model_validator(mode="after")
def validate_emergency_contact(cls, model):
    if model.age > 60 and model.emergency is None:
        raise ValueError("No emergency contact")
    
    return model
```

## computed Filed - can help to calculate sth from fields!


```python
from pydantic import computed_field

@computed_field
@property
def bmi(self)->float: # name of the function is the name of the field
    bmi = round(self.weight/(self.height**2), 2)
    return bmi
```
