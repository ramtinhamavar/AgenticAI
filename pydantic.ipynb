```python
from pydantic import BaseModel
from typing import List, Dict
```

```python
# define the schema!
class SubjectData(BaseModel):
    name: str
    folder_addr: str
    number_of_lost_teeth: int

    annotators: List[str]
    opg_center: Dict[str, str]
```

```python
def add_subject_data(subject: SubjectData):
    print(f"Subject name: {subject.name}")
    print(f"Subject folder addr: {subject.folder_addr}")
    print(f"Subject number of lost teeth: {subject.number_of_lost_teeth}")
    print(f"List of annotators: {subject.annotators}")
    print(f"opg center name: {subject.opg_center.keys()}, opg center code: {subject.opg_center.values()}")
```

```python
subject_data = {"name":"Ramtin", "folder_addr":"/mnt/dev0/ramtinhamavar", "number_of_lost_teeth":2,
               "annotators":["annot_37", "annot_21"], "opg_center":{"Dr. Zhivago":"021"}}
subject = SubjectData(**subject_data)

add_subject_data(subject)
```
